# 9. Autoavaliação do Projeto

## 9.1 Objetivo da autoavaliação

Esta seção apresenta uma reflexão sobre o desenvolvimento do projeto de Engenharia de Dados, considerando os objetivos alcançados, os principais desafios encontrados, os aprendizados obtidos, as limitações da solução desenvolvida e possíveis evoluções futuras.

A autoavaliação considera todo o processo de construção do Lakehouse, desde a exploração e ingestão dos dados históricos da ANP até a aplicação da arquitetura Medalhão, avaliação da qualidade dos dados, transformação e consolidação da camada Silver, modelagem dimensional, construção da camada Gold, documentação do catálogo de dados e realização das análises das perguntas de negócio.

## 9.2 Objetivos alcançados

O objetivo principal do projeto foi construir um pipeline de dados capaz de integrar, tratar, organizar e modelar os dados históricos de produção offshore de petróleo e gás natural no Brasil entre 2005 e 2025, possibilitando sua utilização em análises sobre evolução, distribuição geográfica e concentração da produção.

Considera-se que esse objetivo foi alcançado. Os arquivos históricos disponibilizados pela ANP foram integrados em uma arquitetura Lakehouse organizada segundo o modelo Medalhão, preservando os dados de origem e promovendo sua transformação progressiva nas camadas Bronze, Silver e Gold.

Durante esse processo, foram implementadas etapas de ingestão, avaliação da qualidade, padronização de estrutura e tipos de dados, consolidação histórica, modelagem dimensional e criação de tabelas analíticas em formato Delta.

O resultado final permitiu analisar os 21 anos de dados disponíveis, entre janeiro de 2005 e dezembro de 2025, e responder às quatro perguntas de negócio definidas para o projeto.

Além da construção técnica do pipeline, buscou-se documentar as decisões tomadas ao longo do desenvolvimento, permitindo compreender não apenas as transformações realizadas, mas também as justificativas para os tratamentos adotados.

## 9.3 Principais desafios e aprendizados

O desenvolvimento do projeto apresentou desafios relacionados principalmente à integração de arquivos históricos produzidos em diferentes períodos e à necessidade de garantir que as transformações realizadas não alterassem indevidamente os dados de origem.

Um dos primeiros desafios foi a **heterogeneidade dos arquivos disponibilizados pela ANP**. Embora os arquivos apresentassem os mesmos 21 atributos lógicos, foram identificadas diferenças de encoding, tipos inferidos e formatação dos cabeçalhos entre os períodos. Os arquivos mais recentes, por exemplo, apresentavam os nomes das colunas entre colchetes, enquanto alguns arquivos anteriores exigiam tratamento específico de codificação de caracteres.

Um desafio particularmente relevante ocorreu durante a ingestão dos arquivos de **2019, 2020 e 2021**. A análise inicial revelou comportamentos anômalos nas cardinalidades de atributos categóricos nesses períodos. A investigação mostrou que parte das linhas possuía uma estrutura específica de aspas externas e aspas internas duplicadas, interferindo no parsing dos registros. A solução exigiu investigar diretamente os arquivos de origem, identificar o padrão responsável pelo problema e aplicar uma normalização específica antes da leitura.

Esse processo representou um dos principais aprendizados do projeto: **a quantidade de registros preservados não é suficiente, isoladamente, para garantir a qualidade de uma ingestão**. Mesmo quando a quantidade de linhas entre a origem e a camada Bronze era compatível, alguns valores podiam ter sido interpretados incorretamente. A utilização de validações adicionais, como cardinalidade de atributos categóricos e inspeção dos dados de origem, foi fundamental para identificar o problema.

A etapa de qualidade também reforçou a importância de **diferenciar uma característica do dado de um erro que deve ser corrigido**. Valores nulos não foram automaticamente convertidos em zero, variações históricas na quantidade de campos e instalações não foram classificadas como inconsistências sem evidência e três valores negativos encontrados na produção de água foram preservados após a confirmação de que já estavam presentes nos arquivos de origem.

Outro aprendizado relevante esteve relacionado à **separação de responsabilidades entre as camadas da arquitetura Medalhão**. A camada Bronze preservou os dados ingeridos com o mínimo de transformação necessário para sua leitura, enquanto a Silver concentrou as padronizações de estrutura, tipos e valores categóricos. A camada Gold, por sua vez, foi destinada à modelagem dimensional e às transformações voltadas ao consumo analítico, incluindo as conversões das medidas de produção para unidades mais adequadas às análises.

Por fim, a construção do projeto permitiu aplicar de forma integrada conceitos de Engenharia de Dados como ingestão, armazenamento em Delta, qualidade de dados, transformação, arquitetura Medalhão, modelagem dimensional, chaves substitutas, catálogo de dados, linhagem e construção de uma camada analítica orientada às perguntas de negócio.

## 9.4 Limitações do projeto

Apesar de permitir responder às perguntas de negócio propostas, o projeto apresenta algumas limitações relacionadas ao escopo e às características dos dados utilizados.

A análise foi construída a partir dos dados históricos de produção offshore selecionados para o projeto e, portanto, está limitada aos atributos disponibilizados nessas fontes. Informações externas capazes de ampliar a interpretação dos resultados não foram incorporadas diretamente ao modelo dimensional.

Uma limitação relevante observada durante as análises está relacionada ao **pré-sal**. A expansão do pré-sal apresenta correspondência temporal com algumas mudanças identificadas nos dados, especialmente com o crescimento da Bacia de Santos. Entretanto, a base utilizada no projeto não classifica diretamente os registros analisados entre pré-sal e pós-sal. Dessa forma, informações oficiais externas foram utilizadas apenas para contextualizar os resultados, sem atribuir causalidade a partir dos dados do projeto.

Além disso, as análises de concentração utilizaram a participação dos **cinco e dez maiores campos produtores** como uma medida simples e interpretável da distribuição da produção. Essa abordagem permite responder à pergunta de negócio proposta, mas não corresponde a um índice estatístico formal de concentração.

Por fim, o projeto priorizou a construção completa do pipeline de Engenharia de Dados e a resposta às quatro perguntas de negócio definidas. Assim, outras perspectivas analíticas que poderiam ser exploradas a partir da base foram mantidas fora do escopo para preservar o foco da solução.

## 9.5 Possíveis melhorias e trabalhos futuros

A arquitetura desenvolvida permite diferentes possibilidades de evolução do projeto, tanto na ampliação das fontes de dados quanto na automação do pipeline e no desenvolvimento de análises mais avançadas.

Uma das principais melhorias seria incorporar **dados oficiais que permitam identificar diretamente a produção proveniente do pré-sal**. Essa informação poderia ser integrada ao modelo analítico para avaliar quantitativamente a participação do pré-sal na evolução da produção de óleo e gás natural, permitindo testar diretamente sua relação com as mudanças observadas ao longo da série histórica, em vez de utilizá-la apenas como contexto externo para interpretação dos resultados.

Outra possibilidade seria incorporar **dados sobre reservas de petróleo e gás natural e sua distribuição geográfica**. A análise conjunta das reservas, bacias, campos e localidades poderia permitir avaliar se existem regiões com potencial para ampliar sua participação na produção offshore brasileira e investigar, com dados adicionais, a possibilidade de redução futura da concentração atualmente observada em poucos campos e regiões produtoras.

Outra evolução relevante seria a **automatização do processo de ingestão de novos dados publicados pela ANP**. O pipeline poderia ser adaptado para identificar e carregar automaticamente novos arquivos ou períodos disponibilizados pela fonte, incorporando-os ao Data Lake e executando de forma sequencial as etapas de ingestão, validação, transformação e atualização das camadas Bronze, Silver e Gold. Dessa forma, a solução poderia evoluir de um processamento histórico para um pipeline atualizado periodicamente com novos dados.

Com a ampliação da série histórica e a automatização das atualizações, os dados também poderiam ser utilizados em aplicações de **Machine Learning voltadas à previsão da produção**. Modelos de séries temporais ou outros métodos preditivos poderiam ser avaliados para estimar a evolução futura da produção de óleo e gás natural, considerando diferentes níveis de agregação, como produção total, bacia ou campo.

A combinação de informações históricas de produção, reservas e características das diferentes regiões produtoras também poderia ampliar essas análises preditivas, permitindo investigar possíveis mudanças futuras na distribuição geográfica e na concentração da produção.

Essas evoluções permitiriam transformar a solução desenvolvida neste projeto em uma plataforma de dados continuamente atualizada e capaz de suportar não apenas análises históricas, mas também análises preditivas e novas perspectivas de negócio.
